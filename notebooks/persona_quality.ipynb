# Persona Quality Checks

1. Lade harmonisierte Personas (`data/processed/<source_id>/<batch>/personas.jsonl`).
2. Berechne Coverage-Metriken (Segments, Regionen, Provenance-Anteile).
3. Prüfe Embedding-Drift: cosine distance zu Vorwoche < 0.2.
4. Bias-Scan: demografische Attribute vs. Anteil Datensätze.
5. Exportiere Findings als Markdown und verlinke Audit-Dateien.
